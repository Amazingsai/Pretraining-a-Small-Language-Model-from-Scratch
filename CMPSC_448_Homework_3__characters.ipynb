```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from datasets import load_dataset

# Making the results more repeatable
np.random.seed(10)
tf.random.set_seed(10)
```

```python
from huggingface_hub import login
login()
```

```python
from datasets import load_dataset

dataset = load_dataset("GEM/viggo",trust_remote_code=True)

print(dataset)
print("Available columns:", dataset["train"].column_names)
print("Example:", dataset["train"][0])
```

```python
# Collecting the text from the target column
texts = []

for row in dataset["train"]:
  sentance = row["target"]
  if sentance != "":
    texts.append(sentance)
texts = texts[:1500]
text = "\n".join(texts)
```

```python
# Turning characters into numbers
# each character as an individual token
# uppercase and lowercase characters are distinct characters.
# does not to remove the usual filtered punctuation characters.

tokenizer = Tokenizer(char_level = True, lower = False, filters = "")

```

```python
#Training the tokenizer using text from the target column
tokenizer.fit_on_texts([text])
```

```python
#converting the text into a sequence of numeric IDs.
encoded_text = tokenizer.texts_to_sequences([text])[0]
```

```python
#counts the characters in the vocabulary and adds one because Keras reserves ID 0 for padding
vocab_size = len(tokenizer.word_index) + 1
```

```python
print("Number of different characters:", vocab_size - 1)
print("First 30 character IDs:", encoded_text[:30])
```

```python
#creating training samples
#the model sees 40 characters and learns to predict the next character.
sequence_length = 40
step = 3

x = []
y = []

for i in range(0, len(encoded_text) - sequence_length, step):
  x.append(encoded_text[i:i + sequence_length])
  y.append(encoded_text[i + sequence_length])

x = np.array(x)
y = np.array(y)

print("Input examples shape:", x.shape)
print("Target examples shape:", y.shape)
```

```python
#Building th RNN model
model = tf.keras.Sequential([
    tf.keras.layers.Embedding(
        input_dim=vocab_size,
        output_dim=32
    ),
    tf.keras.layers.SimpleRNN(64),
    tf.keras.layers.Dense(
        vocab_size,
        activation="softmax"
    )
])
```

```python
#Compiling the model
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
#Training the model
history = model.fit(x,y,epochs=10,batch_size=64,validation_split=0.1)
```

```python
# Generating text from a prompt

#Reversing the tokenizer's dictionary character-to-number mapping
#so the program can convert numbers back into characters
id_to_char = {
    number: character
    for character, number in tokenizer.word_index.items()
}

#To generate text, the model predicts a character, then sees the updated text, and predicts another character.
def generate_text(prompt, number_of_characters=200):
    generated = prompt

    for _ in range(number_of_characters):
      # Keeping the most recent characters, and turning them into IDs
      input_ids = tokenizer.texts_to_sequences([generated[-sequence_length:]])[0]


      # Padding by adding zero IDs at the beginning if the prompt is shorter than 40 characters
      if len(input_ids) < sequence_length:
        input_ids = ([0] * (sequence_length - len(input_ids))+ input_ids)

      #Making the prediction
      #The input IDs are converted into an array and passed through the trained model.
      #producing probabilities for the possible next characters.
      input_array = np.array([input_ids])
      predictions = model.predict(input_array, verbose=0)[0]

      # Adjusting the probabilities
      #Using temperature sampling to control how predictable or random the next character will be when the model generates text
      temperature = 0.5
      predictions = np.log(predictions + 1e-8) / temperature
      predictions = np.exp(predictions)
      predictions = predictions / np.sum(predictions)

      # Randomly chooses the next character for the language model to generate.
      next_id = int(np.random.choice(len(predictions), p=predictions))

      #Converting the ID back to text
      #If the predicted ID is invalid for this mapping, the loop stops.
      #But otherwise, the program converts the ID into a character and appends it to the generated text.
      if next_id == 0 or next_id not in id_to_char:
        break

      generated += id_to_char[next_id]

    #This returns the original prompt plus the generated continuation.
    return generated



```

```python
# Try a few different prompts
print("PROMPT 1")
print(generate_text("The character is the best", 200))
print()

print("PROMPT 2")
print(generate_text("I think I like", 200))
print()

print("PROMPT 3")
print(generate_text("This game", 200))

print("PROMPT 4")
print(generate_text("The game", 200))
```
